```python
# Уменьшение длины ряда временных периодов 
times = ["01.30 - 03.00", "03.00 - 4.30", "04.30 - 5.00", "05.00 - 7.00", "07.00 - 7.30", "13.30 - 15.00", "19.00 - 19.30"]

rows = ""
for t in times:
    if len(rows) == 0:
        rows += f"{t}<br>"
    else:
        endp = int(rows.rstrip("<br>").split(" - ")[-1].split(".")[0])
        nextp = int(t.split(" - ")[0].split(".")[0])
        if endp == nextp:
            rows = rows.replace(rows.rstrip("<br>").split(" - ")[-1], t.split(" - ")[1])
        else:
            rows += f"{t}<br>"
            
print(rows)
```

```python
from datetime import datetime, timedelta
from dateutil import easter  # нужно установить: pip install python-dateutil

def get_100_workdays_with_holidays(year=2026, start_date=None):
    """
    Список рабочих дней с учетом праздников РФ
    (упрощенная версия, включает основные праздники)
    """
    
    # Фиксированные праздники РФ
    holidays_fixed = {
        (1, 1): "Новый год",
        (1, 2): "Новогодние каникулы",
        (1, 3): "Новогодние каникулы",
        (1, 4): "Новогодние каникулы",
        (1, 5): "Новогодние каникулы",
        (1, 6): "Новогодние каникулы",
        (1, 7): "Рождество",
        (1, 8): "Новогодние каникулы",
        (2, 23): "День защитника Отечества",
        (3, 8): "Международный женский день",
        (5, 1): "Праздник Весны и Труда",
        (5, 9): "День Победы",
        (6, 12): "День России",
        (11, 4): "День народного единства",
    }
    
    def is_holiday(date):
        """Проверяет, является ли день праздничным"""
        return (date.month, date.day) in holidays_fixed
    
    def is_weekend(date):
        """Проверяет, является ли день выходным"""
        return date.weekday() >= 5  # 5=суббота, 6=воскресенье
    
    if start_date is None:
        start_date = datetime(year, 1, 1).date()
    
    workdays = []
    current_date = start_date
    
    while len(workdays) < 200:
        # Проверяем: не выходной И не праздник
        if not is_weekend(current_date) and not is_holiday(current_date):
            workdays.append(current_date.strftime('%d.%m.%Y'))
        current_date += timedelta(days=1)
    
    return workdays

# Использование
workdays = get_100_workdays_with_holidays(2026, datetime(2026, 4, 1).date())
# for i, day in enumerate(workdays[:10], 1):
#     print(f"{i:3}. {day.strftime('%d.%m.%Y')} ({day.strftime('%A')})")
workdays
```
